**Question (B)** Defining a variable through a piecewise-defined expression

<font size = "3">

Let $I$ denote the income of a single person, measured in thousands of dollars (so if $I = 100$, the person made $100,000). Let $T$ denote the amount of income tax that person paid, again in thousands of dollars. The amount of tax is determined by the following progressive tax bracket:


$$
T = 
\begin{cases}
0.1 \times I & \text{if }  0 \leq I \leq 12\\[6pt]
1.2 + 0.12 \times (I - 12) & \text{if } 12 < I \leq 48.5 \\[6pt]
5.58 + 0.22 \times (I - 48.5) & \text{if } 48.5 < I \leq 103.5 \\[6pt]
17.68 + 0.24 \times (I - 103.5) & \text{if } 103.5 < I \leq 197.5 \\[6pt]
40.24 + 0.32 \times (I - 197.5) & \text{if } 197.5 < I
\end{cases}
$$

In the code cell below, a variable called `income_in_thousands` is randomly generated between $0$ and $400$. Using `if`, `elif`, and/or `else` statements, write code that defines a variable `tax_paid` that is determined by the expression above.

````python
import random

income_in_thousands = round(random.uniform(0, 400), 3)

I = income_in_thousands

tax_paid = 0

# your code here
if (0 <= I and I <= 12): tax_paid = 0.1 * I
elif (12 < I and I <= 48.5): tax_paid = 1.2 + 0.12 * (I - 12)
elif (48.5 < I and I <= 103.5): tax_paid = 5.58 + 0.22 * (I - 48.5)
elif (103.5 < I and I <= 197.5): tax_paid = 17.68 + 0.24 * (I - 103.5)
else: tax_paid = 40.24 + 0.32 * (I - 197.5)

print(I)
print(tax_paid)
````

**Question (C)** Conditional statements

<font size = "3">

The code cell below generates a random integer between 1 and 20 (both 1 and 20 are possible values). Write code that tests the following:

- Tests if the number satisfies $10 < n < 15$

````python
import random

lo = 1
hi = 20

random_int = random.randint(lo, hi)
print(random_int)

# Your code here
n = random_int

if (1 <= n and n < 5): print("The number is between 1 (inclusive) and 5 (exclusive)")
elif (5 <= n and n <= 10): print("The number is between 5 (inclusive) and 10 (inclusive)")
elif (10 < n and n < 15): print("The number is between 10 (exclusive) and 15 (exclusive)")
else: print("The number is greater than or equal to 15")
````

**Question (F)** Categorizing movies


<font size = "3">

Below, two lists are created. One contains the name of movies, and the other the corresponding year it was released ("movie_names[k]" was released in the year "movie_years[k]"). We want to sort them into three categories:

- Pre-1960s: Movies released in 1959 or earlier.

- Latter Twentieth Century: Movies released between 1960 and 1999 (both inclusive).

- Modern movies: Movies released in 2000 or later.

Perform this task using a "for" loop. After the loop, `pre_1960s` should contain the **names** of all movies released 1959 or earlier (Häxan, Citizen Kane, Some Like It Hot). The lists `latter_20th` and `modern_movies` should be similar, containing the names of all movies released in that time period.

````python
movie_names = ["Häxan", "Scream", "Citizen Kane", "Past Lives", "Eraserhead",
               "Mother Joan of the Angels", "Le Samouraï", "A Tale of Two Sisters",
               "Mulholland Drive", "Some Like It Hot"]
movie_years = [1922, 1996, 1941, 2023, 1977, 1961, 1967, 2003, 2001, 1959]

# Each string from "movie_names" should be added to one of these three lists.
pre_1960s = []
latter_20th = []
modern_movies = []

# Write your for loop here:

for i, movie in enumerate(movie_names):
    MY = movie_years[i]
    if (MY < 1960):
        pre_1960s.append(movie)
    elif (1960 <= MY and MY <= 1999):
        latter_20th.append(movie)
    else:
        modern_movies.append(movie)

print(pre_1960s)
print(latter_20th)
print(modern_movies)
````
